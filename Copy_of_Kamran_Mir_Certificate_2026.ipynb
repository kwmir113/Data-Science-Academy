```python
# 1. Loading CSV data from the URL below
import pandas as pd

url = 'https://docs.google.com/spreadsheets/d/e/2PACX-1vRyoYQ2QDFmRZVgAJVQdAzQ0ZQQ7WoRgVrzqdGeGqLN_37T9TXxYzEdImfdFAdXU9Ky73q-K88bmfTc/pub?gid=0&single=true&output=csv'
df = pd.read_csv(url)
```

```python
# 15. Shipping cost under 10, SellerFeedback > 95
df['SellerFeedback'] = pd.to_numeric(df['SellerFeedback'], errors='coerce')
filtered_df = df[(df['ShippingCost'] <= 10) & (df['SellerFeedback'] > 95)]
filtered_df
```

```python
# 16 Rename a column
df = df.rename(columns={'SellerFeedback': 'ReviewRating'})
df.columns.tolist()
```

```python
# 17 Calculate Total Sales for a Category
category_sales = df.groupby('Category')['Quantity'].sum()
print(category_sales)
```

```python
# 17 Calculate total number of Sales in a Category
Category_Total = df[df['Category'] == 'Home Improvement']['Quantity'].sum()
print(f"Total Home Improvement items: {Category_Total}")
```

```python
# 18 Find out average Shipping Cost
avg_shipping_cost = df['ShippingCost'].mean()
print(f"Average Shipping Cost: ${avg_shipping_cost:.2f}")
```

```python
#19 Sum and Count Prices for Beauty Products from China
sum_beautyProduct_China = df[(df['Category'] == 'Beauty') & (df['SellerCountry'] == 'China')]['Price'].sum()

print(f"Total Price for Beauty products from China: ${beauty_china_sum:.2f}")
```

```python
#19 Sum and Count Prices for Beauty Products from China
sum_beautyProduct_China = df[(df['Category'] == 'Beauty') & (df['SellerCountry'] == 'China')]['Price'].sum()

print(f"Total Price for Beauty products from China: ${beauty_china_sum:.2f}")
```

```python
#19 Sum and Count Prices for Beauty Products from China
sum_beautyProduct_China = df[(df['Category'] == 'Beauty') & (df['SellerCountry'] == 'China')]['Price'].sum()

print(f"Total Price for Beauty products from China: ${sum_beautyProduct_China:.2f}")
```

```python
# 20 Add a new Column with Null values
import numpy as np
df['NewColumn'] = np.nan
df.isna().sum()
```

```python
# 21 List average price by sub category
electronics_avg_price = df[df['Category'] == 'Electronics'].groupby('SubCategory')['Price'].mean().reset_index()
electronics_avg_price.columns = ['SubCategory', 'AveragePrice']
print(electronics_avg_price)
```

```python
# 22 Filter by Delivery time and quantity
filtered_data = df[(df['DeliveryTime'] <= 5) & (df['Quantity'] > 10)]
# Display the filtered rows
filtered_data
```

```python
# Find the Maxiumu price among all products in the dataset
max_price = df['Price'].max()

print(f"Maximum price among all products: ${max_price:.2f}")
```

```python
# 24 Find Total Sales per Seller
df['SalesAmount'] = df['Price'] * df['Quantity']
seller_sales = df.groupby('SellerId')['SalesAmount'].sum().reset_index()
# Display the results
print(seller_sales)
```

```python
# 25 Filter products with "Shoes" in the description
shoes_products = df[df['Description'].str.contains('Shoes', case=False, na=False)]

# Display the filtered rows
shoes_products
```
